```python
# import pySpark
# it is an entry point for working with structured data using Spark SQL
from pyspark.sql import SparkSession

# Create a spark session object named spark using the Build pattern
# appName set the name of the spark application
# getOrCreate method either retrieves an existing spark session or 
# create a new one if it doesn't exists.
spark = SparkSession.build.appName("PySpark-Get-Started").getOrCreate()

# Test the setup by using simple data rame to create data
data = [{"Alice", 25), ("Bob", 30), ("Charlie",35)]
df = spark.CreateDataFrame(data, ["Name","Age"])
df.show()
                                   
```
